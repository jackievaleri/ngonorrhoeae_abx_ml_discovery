# Compare model performances

Overall Goal
- Aggregate held-out test performance (auPR and auROC) across all 2D models trained on the PK + 37K screen: RFC, SVM, FFN, ChemBERTa, AttentiveFP, and D-MPNN (Chemprop).
- Visualize per-fold performance distributions side-by-side under both random and scaffold splits.
- Statistically compare the top-performing D-MPNN scaffold model against all alternatives to justify its selection for downstream prediction.

Relation to Manuscript
- Related to Figure 2D.

Takeaways
- D-MPNN (Chemprop) achieves higher auROC than FFN, SVM, ChemBERTa, and AttentiveFP, and is statistically indistinguishable from RFC.
- Scaffold-split performance is consistently lower than random-split performance, confirming the importance of scaffold splits for honest generalization estimates.
- The classical RFC baseline remains competitive on auROC.

Helper Functions
- `extend_dataframe`: Appends per-fold metric values for a given model into a long-format DataFrame suitable for seaborn plotting.

Scientific Methodology
- Loaded per-fold test-set scores from each model's results directory (`summary_results.csv` for ChemBERTa/AttentiveFP, `test_scores.csv` for RFC/SVM/FFN/D-MPNN).
- Combined results into a single long-format DataFrame indexed by (Value, Metric, Model) with both random- and scaffold-split variants per model.
- Used a `CategoricalDtype` to enforce a consistent model ordering with spacer ("dummy") entries for visual grouping.
- Plotted overlaid swarm + bar plots (per-fold points over mean bars) for auPR and auROC, with hatching to distinguish random vs scaffold splits, saved as SVG and PNG figure panels.
- Performed independent two-sample t-tests comparing D-MPNN:Scaffold against each other model's scaffold-split distribution for both metrics.
- Exported the underlying combined DataFrame to `../figure_panels/underlying_data/model_comparison_values.csv` for manuscript reproducibility.


## Setup

```python
import matplotlib.pyplot as plt
import pandas as pd
import scipy.stats as sp
import seaborn as sns
from IPython.display import display
from pandas.api.types import CategoricalDtype
```

## Helper functions

```python
def extend_dataframe(olddf, vals, metric, name):
    """
    Extends an existing DataFrame with new evaluation results.

    Creates a temporary DataFrame containing values, their associated metric,
    and model name, then appends it to the input DataFrame.

    Parameters:
    olddf (pd.DataFrame): Existing DataFrame to extend.
    vals (list or array-like): List of numeric values (e.g., AUROC, AUPR).
    metric (str): Name of the evaluation metric (e.g., 'AUROC', 'AUPR').
    name (str): Name of the model producing the results.

    Returns:
    pd.DataFrame: Updated DataFrame containing both old and new rows with columns:
        - 'Value': Numeric values from vals.
        - 'Metric': Metric name repeated for each value.
        - 'Model': Model name repeated for each value.
    """
    new = pd.DataFrame()
    new["Value"] = vals
    new["Metric"] = metric
    new["Model"] = name
    olddf = olddf.append(new)
    return olddf
```

## Make plots

```python
newdf = pd.DataFrame(columns=["Value", "Metric", "Model"])

labels = [
    "RFC:Random",
    "RFC:Scaffold",
    "SVM:Random",
    "SVM:Scaffold",
    "FFN:Random",
    "FFN:Scaffold",
    "dummy1",
    "ChemBERTa:Random",
    "ChemBERTa:Scaffold",
    "dummy2",
    "AttentiveFP:Random",
    "AttentiveFP:Scaffold",
    "dummy3",
    "D-MPNN:Random",
    "D-MPNN:Scaffold",
]

# paths since they are all in different folders
rfc_rand_path = "rfc_hyperopt_pk_37k/rfc_RANDOM_final_122/"
rfc_path = "rfc_hyperopt_pk_37k/rfc_final_122/"
svm_rand_path = "svm_hyperopt_pk_37k/svm_RANDOM_final_15/"
svm_path = "svm_hyperopt_pk_37k/svm_final_15/"
ffn_rand_path = "ffn_hyperopt_pk_37k/ffn_RANDOM_final_20/"
ffn_path = "ffn_hyperopt_pk_37k/ffn_final_20/"

chemb_prefix = "chemberta_pk_37k/"
chemb_rand_path = chemb_prefix + "random_split_chemberta_models/"
chemb_scaff_path = chemb_prefix + "scaffold_split_chemberta_models/"

attent_prefix = "attentivefp_pk_37k/"
attent_rand_path = attent_prefix + "random_split_attentivefp_models/"
attent_scaff_path = attent_prefix + "scaffold_split_attentivefp_models/"

gnn_prefix = "pk_37k_screen_models_03192022/"
gnn_rand_path = (
    gnn_prefix + "FINALbayHO04052022_30_models_no_scaffold_split/"
)  # 30 individual folds for fairer comparison
gnn_scaff_path = (
    gnn_prefix + "FINALbayHO04052022_30_models_with_scaffold_split/"
)  # 30 individual folds for fairer comparison
for fold, name in zip(
    [
        rfc_rand_path,
        rfc_path,
        svm_rand_path,
        svm_path,
        ffn_rand_path,
        ffn_path,
        "",
        chemb_rand_path,
        chemb_scaff_path,
        "",
        attent_rand_path,
        attent_scaff_path,
        "",
        gnn_rand_path,
        gnn_scaff_path,
    ],
    labels,
):
    if "dummy" in name:
        newdf = extend_dataframe(newdf, [-0.1], "auPR", name)  # empty
        newdf = extend_dataframe(newdf, [-0.1], "auROC", name)

    elif "ChemBERTa" in name or "AttentiveFP" in name:
        df = pd.read_csv("../models/other_models/" + fold + "summary_results.csv")
        newdf = extend_dataframe(newdf, list(df["auPR"]), "auPR", name)  # 20 folds
        newdf = extend_dataframe(newdf, list(df["auROC"]), "auROC", name)

    else:
        if "D-MPNN" not in name:
            df = pd.read_csv("../models/other_models/" + fold + "test_scores.csv")
        else:
            df = pd.read_csv("../models/" + fold + "test_scores.csv")
        newdf = extend_dataframe(newdf, list(df.iloc[0, 3:33]), "auPR", name)  # 30 folds
        newdf = extend_dataframe(newdf, list(df.iloc[0, 35:65]), "auROC", name)

    # previously needed this for the case where D-MPNN had 5 folds of 10 ensemble models each
    # now retrain with 30 individual folds like RFC, SVM, FFN
    # else:
    #     print(name)
    #     df = pd.read_csv('../models/' + fold + 'test_scores.csv')
    #     newdf = extend_dataframe(newdf, list(df.iloc[0,3:8]), 'auPR', name) # 5 folds of 10 ensemble models each
    #     newdf = extend_dataframe(newdf, list(df.iloc[0,10:15]), 'auROC', name)

df = newdf
df
```

```python
# new order so GNN is last
cat_model_order = CategoricalDtype(labels, ordered=True)
df["Model"] = df["Model"].astype(cat_model_order)
df.sort_values("Model")
df
```

```python
plt.figure(figsize=(5, 4), dpi=300)
aupr_df = df[df["Metric"] == "auPR"]
my_pal = dict(
    zip(
        labels,
        [
            "lightgrey",
            "lightgrey",
            "lightslategrey",
            "lightslategrey",
            "slategrey",
            "slategrey",
            "white",
            "dimgrey",
            "dimgrey",
            "white",
            "darkgrey",
            "darkgrey",
            "white",
            "royalblue",
            "royalblue",
        ],
    )
)
ax = sns.swarmplot(
    data=aupr_df,
    x="Model",
    y="Value",
    edgecolor="black",
    linewidth=0.2,
    s=2,
    label="",
    palette=my_pal,
)
sns.barplot(
    data=aupr_df,
    x="Model",
    y="Value",
    edgecolor="black",
    linewidth=1,
    alpha=0.7,
    palette=my_pal,
)
handles, labs = ax.get_legend_handles_labels()

# Define some hatches
# Loop over the bars
# weird numbering convention - loops over colors
# https://stackoverflow.com/questions/35467188/is-it-possible-to-add-hatches-to-each-individual-bar-in-seaborn-barplot # noqa: E501
# extend_labels = [x for x in labels for _ in (0, 1)]
# # can use for plotting auPR and auROC in same plot
hatches = [" " if "Scaffold" in x else "\\\\\\" for x in labels]

for i, thisbar in enumerate(ax.patches):
    thisbar.set_hatch(hatches[i])

plt.xlabel("Model")
plt.ylim([0, 1.0])
plt.xticks(rotation=80)
plt.ylabel("Area Under P-R Curve")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()

plt.savefig("../figure_panels/model_comparison_aupr.svg")
plt.savefig("../figure_panels/model_comparison_aupr.png")
plt.show()
```
